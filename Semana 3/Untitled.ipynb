```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import squareform

# Definir la matriz de distancias
dist_matrix = np.array([
    [0,    0.12, 0.51, 0.84, 0.28, 0.34],
    [0.12, 0,    0.25, 0.16, 0.77, 0.61],
    [0.51, 0.25, 0,    0.14, 0.70, 0.93],
    [0.84, 0.16, 0.14, 0,    0.45, 0.2],
    [0.28, 0.77, 0.70, 0.45, 0,    0.67],
    [0.34, 0.61, 0.93, 0.2, 0.67, 0   ]
])

# Convertir a formato condensed
dist_condensed = squareform(dist_matrix)

# Calcular linkage (puedes cambiar 'average' por 'single', 'complete', 'ward')
Z = linkage(dist_condensed, method='complete')

# Graficar dendrograma
plt.figure(figsize=(7, 4))
dendrogram(Z, labels=["A", "B", "C", "D", "E", "F"])
plt.title("Dendrograma de puntos A-F")
plt.xlabel("Puntos")
plt.ylabel("Distancia")
plt.show()
```
