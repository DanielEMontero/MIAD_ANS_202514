### 2. Homogenización de textos

Para cumplir con el objetivo de generar recomendaciones en esta sección debe preparar los posts para poder ser utilizados en su sistema de recomendación. Para ello, "limpie" y "tokenize" cada uno de los cuentos, describiendo detalladamente los pasos que realizo y si transformó o eliminó ciertas palabras. Para asistirlo en la tarea he creado listas de *stopwords* que están disponibles en la carpeta `data`. En su procedimiento ilustre la limpieza con el cuento 'La venganza del metegol'. (En su limpieza recuerde que el objetivo es generar recomendaciones a partir de la similitud de las palabras o temas de los cuentos)

```python
# Utilice este espacio para escribir el código.
# helper display
def display_clean(df:pd.DataFrame)->None:
    """Muestra las primeras n filas del DataFrame y su información.

    Args:
        df (pd.DataFrame): DataFrame a mostrar.
        n (int, optional): Número de filas a mostrar. Por defecto es 5.
    """
    print("Palabras-Caracteres:", len(df.values[0][0]))
    display(df)

print('===== Texto Crudo ======')
title: str = "La venganza del metegol"
tiny_place = blog_df.loc[blog_df['titulo']==title,['cuento']]
display_clean(tiny_place)


print('===== Texto Limpieza Unicode ======')
blog_df['cuento'] = blog_df['cuento'].apply(unidecode.unidecode)
tiny_place = blog_df.loc[blog_df['titulo']==title,['cuento']]
display_clean(tiny_place)


print('===== Texto Limpieza Caracteres Especiales ======')
blog_df['cuento'] = blog_df['cuento'].apply(lambda x: re.sub(r'[^A-Za-z0-9 ]+', ' ', x))
blog_df['cuento'] = blog_df['cuento'].apply(lambda x: re.sub('\s+', ' ', x))
tiny_place = blog_df.loc[blog_df['titulo']==title,['cuento']]
display_clean(tiny_place)


print('===== Texto Limpieza Espacios Adicionales ======')
blog_df['cuento'] = blog_df['cuento'].apply(lambda x: re.sub("\d+", "", x))
blog_df['cuento'] = blog_df['cuento'].apply(lambda x: re.sub('\s+', ' ',x))
blog_df['cuento'] = blog_df['cuento'].apply(lambda x: x.strip())
tiny_place = blog_df.loc[blog_df['titulo']==title,['cuento']]
display_clean(tiny_place)


print('===== Texto a Minuscula ======')
blog_df['cuento'] = blog_df['cuento'].apply(lambda x: x.lower() if isinstance(x, str) else x)
tiny_place = blog_df.loc[blog_df['titulo']==title,['cuento']]
display_clean(tiny_place)


print('===== Tokenizacion Inicial Spacy ======')
blog_df['documento'] = blog_df['cuento'].apply(lambda x: nlp(x))
tiny_place = blog_df.loc[blog_df['titulo']==title,['documento']]
display_clean(tiny_place)

print('===== Stopword Unicode y Minisculas ======')
extra_stopwords_df[0] = extra_stopwords_df[0]\
    .apply(unidecode.unidecode) \
    .apply(lambda x: x.strip().lower())
display(extra_stopwords_df.head(10))

workshop_stopwords_df[0] = workshop_stopwords_df[0]\
    .apply(unidecode.unidecode) \
    .apply(lambda x: x.strip().lower())
display(workshop_stopwords_df.head(10))   
```

```python
stopwords_self = set(extra_stopwords_df[0].tolist() + workshop_stopwords_df[0].tolist())

# nlp.Defaults.stop_words es un set inmutable ligado al vocabulario, 
# y spaCy lo usa junto con un flag interno en cada Lexeme (lex.is_stop)
nlp.Defaults.stop_words |= stopwords_self
```

```python
# Función para eliminar stopwords
def remove_stopwords(doc):
    return [
        token.text for token in doc 
        if token.text.lower() not in nlp.Defaults.stop_words and not token.is_punct
    ]

# Aplicar a tu columna 'documento'
blog_df["tokens"] = blog_df["documento"].apply(remove_stopwords)
blog_df["tokens"] = blog_df["tokens"].apply(lambda x: " ".join(x))

print('===== Tokenizacion con Extra StopWords ======')
tiny_place = blog_df.loc[blog_df['titulo']==title,['tokens']]
display_clean(tiny_place)


print('===== Tokenizacion Lematizacion ======')
blog_df["tokens"] = blog_df["tokens"].apply(lambda x: " ".join([token.lemma_ for token in nlp(x)]))
tiny_place = blog_df.loc[blog_df['titulo']==title,['tokens']]
display_clean(tiny_place)
```

### 3. Generando Recomendaciones

#### 3.1. Recomendaciones basadas en contenidos

##### 3.1.1. Genere 5 recomendaciones de más recomendada (1) a menos recomendada (5) para el cuento 'La venganza del metegol' usando en la distancia de coseno donde el texto este vectorizado por `CountVectorizer`. Explique el procedimiento que realizó y como ordenó las recomendaciones.

```python
# Utilice este espacio para escribir el código.
count = CountVectorizer()
count_matrix = count.fit_transform(blog_df.tokens)
cosine_sim_count = cosine_similarity(count_matrix, count_matrix)
display(cosine_sim_count)


def recommendation_engine(
    query_title: str,
    cosine_sim,
    df: pd.DataFrame,
    title_col: str = "titulo",
    top_n: int = 5
) -> pd.DataFrame:
    """
    Genera un ranking de recomendaciones basado en similitud coseno.

    Args:
        query_title (str): Título del ítem de referencia.
        cosine_sim (np.ndarray): Matriz de similitudes coseno.
        df (pd.DataFrame): DataFrame con al menos una columna de títulos.
        title_col (str): Nombre de la columna en df que contiene los títulos.
        top_n (int): Número de recomendaciones a devolver.

    Returns:
        pd.DataFrame: Recomendaciones con puesto, título y valor de similitud.
    """
    # Asegurar índice limpio
    df = df.copy().reset_index(drop=True)

    # Mapear títulos a índices
    indices = pd.Series(df.index, index=df[title_col]).drop_duplicates()

    if query_title not in indices:
        raise ValueError(f"El título '{query_title}' no está en la columna '{title_col}'.")

    # Índice del título buscado
    idx = indices[query_title]

    # Similitudes con todos los ítems
    sim_scores = list(enumerate(cosine_sim[idx]))

    # Ordenar por similitud descendente
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Quitar el primero (es el mismo título) y seleccionar top_n
    sim_scores = sim_scores[1 : top_n + 1]

    # Separar índices y valores de similitud
    item_indices = [i[0] for i in sim_scores]
    sim_values = [i[1] for i in sim_scores]

    # Retornar DataFrame con resultados
    return pd.DataFrame({
        "Puesto": range(1, top_n + 1),
        "Recomendación": df[title_col].iloc[item_indices].values,
        "Similitud": sim_values
    })  
    
    
recom_count = recommendation_engine(
    query_title=Constant.TITLE, cosine_sim=cosine_sim_count, df=blog_df, top_n=5
)

display(recom_count)
```

##### 3.1.3. Genere 5 recomendaciones de más recomendada (1) a menos recomendada (5) para el cuento 'La venganza del metegol' usando el texto vectorizado por `TF-IDFVectorizer` y la correlación como medida de similitud.
```python
# Utilice este espacio para escribir el código.
def recommendation_corr(
    query_title: str,
    corr_matrix: pd.DataFrame,
    top_n: int = 5
) -> pd.DataFrame:
    """
    Genera recomendaciones basadas en correlación de Pearson entre documentos.

    Args:
        query_title (str): Documento de referencia.
        corr_matrix (pd.DataFrame): Matriz de correlaciones.
        top_n (int): Número de recomendaciones a devolver.

    Returns:
        pd.DataFrame: Recomendaciones con puesto, título y similitud.
    """
    if query_title not in corr_matrix.columns:
        raise ValueError(f"El título '{query_title}' no está en la matriz de correlación.")

    # Extraer correlaciones con el documento de referencia
    sim_scores = corr_matrix[query_title].drop(query_title)

    # Ordenar de mayor a menor correlación
    sim_scores = sim_scores.sort_values(ascending=False).head(top_n)

    return pd.DataFrame({
        "Puesto": range(Constant.ONE, len(sim_scores) + Constant.ONE),
        "Recomendación": sim_scores.index,
        "Similitud": sim_scores.values
    })

corr_matrix = df_tfidf.T.corr(method="pearson")

recom_tfidf_corr = recommendation_corr(
    query_title=Constant.TITLE,
    corr_matrix=corr_matrix,
    top_n=5
)

display(recom_tfidf_corr)
```

>NOTA:
- Uso `sklearn.decomposition.LatentDirichletAllocation` en lugar de gensim.

- Se convierte los tokens de blog_df.tokens en texto para que CountVectorizer pueda procesarlos.

- Definir una clase Constant con parámetros de configuración (igual estilo que tu código previo).

- Obtener la distribución de temas por documento (lda_topics).

- Identificar el tema principal del cuento objetivo y medimos la probabilidad de otros cuentos en ese tema.

- Generar un ranking de recomendaciones con sus puestos.

```python
# Utilice este espacio para escribir el código.

# Preprocesamiento
text_lda = blog_df.tokens.tolist()

vectorizer = CountVectorizer()
doc_term_matrix = vectorizer.fit_transform(text_lda)

# Modelo LDA con sklearn
lda_model = LatentDirichletAllocation(
    n_components=7,
    random_state=0
)
lda_topics = lda_model.fit_transform(doc_term_matrix)

# Identificar tema principal del cuento
idx_cuento = blog_df[blog_df["titulo"] == Constant.TITLE].index[0]
tema_principal = lda_topics[idx_cuento].argmax()

# Probabilidades de pertenecer al tema principal
prob_tema_principal = lda_topics[:, tema_principal]

# Construir ranking de recomendaciones
df_recom = blog_df[["titulo"]].copy()
df_recom["prob_tema"] = prob_tema_principal

# eliminar el mismo cuento
df_recom = df_recom[df_recom["titulo"] != Constant.TITLE]

# ordenar por mayor probabilidad
df_recom = df_recom.sort_values("prob_tema", ascending=False)

# top N recomendaciones
top_recom = df_recom.head(5).reset_index(drop=True)
top_recom.insert(0, "Puesto", range(1, len(top_recom) + 1))

display(top_recom)
```
